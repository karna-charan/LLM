```python
import torch
import torch.nn.functional as F

# Define query, key, and value tensors
query = torch.tensor([[2.0, 1.0]])
key = torch.tensor([[1.0, 1.0],
                    [2.0, 0.0]])
value = torch.tensor([[5.0, 0.0],
                      [0.0, 5.0]])

# Calculate similarity scores
score = torch.mm(query, key.t())

# Turn scores into attention weights
weights = F.softmax(score, dim=1)

# Compute the final attention output
output = torch.mm(weights, value)

# Show the results
print("Scores:", score)
print("Weights:", weights)
print("Output:", output)
```
